## Schriftelijk examen voorbereiding

### Concepten die je moet begrijpen:

**1. Kepler's 3e wet (circulaire banen)**
- Periode $T$ en radius $r$ hangen samen via: $r = \left(\frac{GM T^2}{4\pi^2}\right)^{1/3}$
- Centrale formule voor initiële condities

**2. Orbitale mechanica**
- Gravitatiekracht: $\vec{F} = -\frac{GM \vec{r}}{r^3}$
- Voor circulaire baan: snelheid $v = \frac{2\pi r}{T}$ (raakt aan baan)
- Energie conservering in gesloten banen

**3. Thrust & Propulsie**
- Thrust verandert snelheid/momentum geleidelijk
- Eenhedenconversie kritisch: N → km/h²
- Optimale burn timing verschuift apogee/perigee

**4. Ellips karakterisering**
- Semi-major as $a$ (halve lange as)
- Semi-minor as $b$ (halve korte as)
- Rotatie hoek $\theta$ w.r.t. x-as
- Linear least squares fit: bepaal $a, b, \theta$ uit trajectory data

**5. Numerieke optimalisatie**
- COBYQA: constrained derivative-free method
- Constraints: ≥2 orbits, burn ≤0.5T, t_start < t_stop
- Cost function gewichten: km² vs graden (factor 100)

### Mogelijke examenvragen + antwoorden:

**V1: Wat is de fysische betekenis van een "burn window"?**
A: Interval [t_start, t_stop] waarin motor brandt. Beperkt door:
- Moet na ≥2 orbits (stabiliteit)
- Niet langer dan 0.5T (brandstofverbruik)

**V2: Waarom gebruiken we linear least squares voor ellips fit?**
A: De vergelijking $x^2 = Ay^2 + Bxy + Cx + Dy + E$ is lineair in parameters. Minimaliseer residue kwadraten → eenduidige oplossing.

**V3: Hoe beïnvloedt throttle timing de orbit?**
A: Vroeg branden → groter effect op apogee. Laat branden → groter effect op perigee. Timing bepaalt welke as groeit.

**V4: Waarom wegen we $\theta$ met factor 100 in cost function?**
A: Dimensionale balans. $a, b$ zijn in km (~10000), $\theta$ in graden (~20). Factor 100 geeft gelijkhang gewicht.

**V5: Wat betekent het als fit_ellipse inf waarden retourneert?**
A: Ongeldige ellips (negatief onder wortel). COBYQA gebruikt dit als penalty → vermijdt dit gebied.
