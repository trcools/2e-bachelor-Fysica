# Analytical Details for the Degradation of Glucose model 

We study the planar ODE system with parameters $a,b>0$:
$$
\begin{aligned}
X' &= -X + aY + X^2Y,\\
Y' &= b - aY - X^2Y.
\end{aligned}
$$

It is convenient to factor the common term $a+X^2$:
$$
\boxed{
\begin{aligned}
X' &= -X + (a+X^2)\,Y,\\
Y' &= b - (a+X^2)\,Y.
\end{aligned}}
$$

---



## 1) Nullclines and qualitative direction field

### $X$-nullcline ($X'=0$)
$$
-X + (a+X^2)Y=0
\quad\Longrightarrow\quad
\boxed{\,Y=\dfrac{X}{a+X^2}\,}.
$$

### $Y$-nullcline ($Y'=0$)
$$
b-(a+X^2)Y=0
\quad\Longrightarrow\quad
\boxed{\,Y=\dfrac{b}{a+X^2}\,}.
$$

### Sign structure (direction arrows in regions)

Because
$$
X'=(a+X^2)\Bigl(Y-\frac{X}{a+X^2}\Bigr),
\qquad
Y'=(a+X^2)\Bigl(\frac{b}{a+X^2}-Y\Bigr),
$$
we get:

- **Above** the $X$-nullcline $Y>\dfrac{X}{a+X^2}$: $X'>0$ (flow points to the right).
- **Below** the $X$-nullcline $Y<\dfrac{X}{a+X^2}$: $X'<0$ (flow points to the left).

- **Above** the $Y$-nullcline $Y>\dfrac{b}{a+X^2}$: $Y'<0$ (flow points downward).
- **Below** the $Y$-nullcline $Y<\dfrac{b}{a+X^2}$: $Y'>0$ (flow points upward).

So the nullclines partition the plane into regions with (right/left) × (up/down) directions.



---

## 2) Equilibrium point(s)

Equilibria satisfy both nullcline equations:
$$
\frac{X}{a+X^2}=\frac{b}{a+X^2}
\quad\Longrightarrow\quad
\boxed{X^{eq} = b}.
$$
Then
$$
Y^{eq}=\frac{b}{a+(b)^2}.
$$

Hence for every $a,b>0$ there is exactly **one** equilibrium:
$$
\boxed{(X^{eq},Y^{eq})=\Bigl(b,\ \frac{b}{a+b^2}\Bigr)}.
$$

---


## 3) Linear stability classification via Jacobian

Let
$$
f(X,Y)=-X+(a+X^2)Y,\qquad g(X,Y)=b-(a+X^2)Y.
$$
Then
$$
J(X,Y)=
\begin{pmatrix}
\partial_X f & \partial_Y f\\
\partial_X g & \partial_Y g
\end{pmatrix}
=
\begin{pmatrix}
-1+2XY & a+X^2\\
-2XY & -(a+X^2)
\end{pmatrix}.
$$

At the equilibrium $(X^{eq},Y^{eq})=(b,\ b/(a+b^2))$, define
$$
S:=a+b^2>0,
\qquad
2X^{eq}Y^{eq}= \frac{2b^2}{S}.
$$
So
$$
J^{eq}=
\begin{pmatrix}
-1+\frac{2b^2}{S} & S\\[4pt]
-\frac{2b^2}{S} & -S
\end{pmatrix}
=
\begin{pmatrix}
\frac{b^2-a}{S} & S\\[4pt]
-\frac{2b^2}{S} & -S
\end{pmatrix}.
$$


### Trace and determinant
$$
\tau := \mathrm{tr}(J^{eq})=\frac{b^2-a}{S}-S,
\qquad
\Delta := \det(J^{eq}) = S = a+b^2.
$$

Key consequence:
$$
\boxed{\Delta=a+b^2>0\ \text{for all}\ a,b>0.}
$$
So the equilibrium is **never a saddle**. Its type is determined by $\tau$ and the discriminant
$$
D := \tau^2-4\Delta.
$$

- Stable if $\tau<0$ (node/focus depending on $D$).
- Unstable if $\tau>0$.
- Hopf threshold when $\tau=0$ (then $D=-4\Delta<0$: purely imaginary eigenvalues).

Eigenvalues explicitly:
$$
\lambda_{\pm}=\frac{1}{2}\Bigl(\tau \pm \sqrt{\tau^2-4\Delta}\Bigr).
$$


### For the given numbers $a=0.06,\ b=0.4$
Compute
$$
S=a+b^2=0.06+0.16=0.22,
$$
$$
\tau=\frac{b^2-a}{S}-S=\frac{0.16-0.06}{0.22}-0.22\approx 0.2345>0.
$$
So the equilibrium is **unstable** (and since typically $D<0$ here, it is an unstable spiral/focus).

---

## 4) Bifurcations when varying $b\in[0,1.2]$ (with fixed $a$)

A local bifurcation of the unique equilibrium happens when $\tau=0$:
$$
\frac{b^2-a}{a+b^2}-(a+b^2)=0.
$$
Let $u=b^2$ so $a+b^2=a+u$. Then $\tau=0$ becomes
$$
\frac{u-a}{a+u}=a+u
\quad\Longleftrightarrow\quad
u-a=(a+u)^2.
$$
Expand:
$$
u-a=a^2+2au+u^2
\quad\Longleftrightarrow\quad
u^2+(2a-1)u+(a^2+a)=0.
$$
Therefore
$$
\boxed{
u=b^2=\frac{1-2a\pm\sqrt{1-8a}}{2}.
}
$$

### Existence condition
Real solutions require
$$
1-8a\ge 0 \quad\Longleftrightarrow\quad \boxed{a\le \tfrac18}.
$$
- If $a>\tfrac18$: no real roots $\Rightarrow \tau<0$ for all $b$ (equilibrium always stable).
- If $0<a\le\tfrac18$: two Hopf points (in $b$) may occur.

Define the Hopf curves (when they exist):
$$
b_\pm(a)=\sqrt{\frac{1-2a\pm\sqrt{1-8a}}{2}},
\qquad b_-(a)\le b_+(a).
$$



### Stability as a function of $b$ (for fixed $a\le 1/8$)
Because $\Delta>0$, stability switches only when $\tau$ changes sign. One finds:
- $\tau<0$ for $0\le b<b_-(a)$: equilibrium stable.
- $\tau>0$ for $b_-(a)<b<b_+(a)$: equilibrium unstable.
- $\tau<0$ for $b>b_+(a)$: equilibrium stable again.

### For $a=0.06$
$$
b_\pm=\sqrt{\frac{1-2(0.06)\pm\sqrt{1-8(0.06)}}{2}}
=\sqrt{\frac{0.88\pm\sqrt{0.52}}{2}}.
$$
Numerically this gives approximately
$$
\boxed{b_-(0.06)\approx 0.282,\qquad b_+(0.06)\approx 0.895.}
$$
So as $b$ increases:
- stable equilibrium for $b<0.282$,
- unstable equilibrium for $0.282<b<0.895$,
- stable equilibrium again for $b>0.895$.

These crossings are **Hopf bifurcations** (since $\tau=0$ and $\Delta>0$ with complex conjugate eigenvalues at threshold).

> Whether each Hopf is supercritical/subcritical (stable vs unstable limit cycle created) is determined by the first Lyapunov coefficient (a higher-order calculation). In practice you usually confirm the cycle's stability numerically, but the *location* of the Hopf curves is fully analytic as above.

---

## 5) Varying $a\in[0,0.14]$: 2D stability diagram in $(a,b)$

The equilibrium is always
$$
(X^{eq},Y^{eq})=\Bigl(b,\frac{b}{a+b^2}\Bigr),
$$
and its stability is determined by the trace sign:
$$
\boxed{
\tau(a,b)=\frac{b^2-a}{a+b^2}-(a+b^2).
}
$$

### Hopf (stability boundary)
The stability boundary is $\tau(a,b)=0$, equivalently
$$
\boxed{
b^2=\frac{1-2a\pm\sqrt{1-8a}}{2},
\qquad 0<a\le \tfrac18.
}
$$

### Interpretation in the $(a,b)$-plane
- For $a>\tfrac18$ (note $0.14>0.125$): **no Hopf** anywhere; the equilibrium is **stable for all $b>0$**.
- For $0<a\le\tfrac18$: there are two Hopf curves $b=b_-(a)$ and $b=b_+(a)$. The equilibrium is:
  - stable below the lower curve,
  - unstable in the band between the curves,
  - stable above the upper curve.

This is exactly the region you would shade in a 2D stability plot: a "tongue/band" of instability for small enough $a$, which closes up and disappears at $a=\tfrac18$.

---

## 6) What you should *notice* in trajectories (qualitative, but explained by the math)

- For parameters where $\tau<0$: trajectories near $(X^{eq},Y^{eq})$ decay to the equilibrium (spiral in if $D<0$, monotone in if $D>0$).
- For parameters where $\tau>0$: trajectories near $(X^{eq},Y^{eq})$ are repelled. In a planar system with a single equilibrium, this commonly leads to an attracting periodic orbit (limit cycle) if trajectories remain bounded; the Hopf bifurcation(s) are the natural mechanism that creates/destroys such a cycle.

A very useful identity for bounding intuition is the sum equation:
$$
(X+Y)'=X'+Y'=(-X+(a+X^2)Y)+(b-(a+X^2)Y)=b-X.
$$
So when $X>b$, the total $X+Y$ decreases; when $X<b$, it increases. This already hints at a "restoring" tendency around $X\approx b$, which is consistent with oscillatory behavior near Hopf regimes.
