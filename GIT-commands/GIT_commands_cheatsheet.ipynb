# Meest gebruikte Git-commands (cheat sheet)

## 1) Starten & repo-info
- `git init` — maak een nieuwe Git-repository in de huidige map.
- `git clone <url>` — kopieer een bestaande repo (incl. geschiedenis) naar je computer.
- `git status` — toon wat er gewijzigd is, wat staged is, en op welke branch je zit.
- `git log` — toon commitgeschiedenis (tip: `--oneline --graph --decorate` voor compact overzicht).
- `git diff` — toon verschillen t.o.v. de laatste commit (unstaged).
- `git diff --staged` — toon verschillen die al staged zijn.

## 2) Changes selecteren (staging) & committen
- `git add <bestand>` — voeg een bestand toe aan de staging area.
- `git add .` — stage alle wijzigingen in de huidige map (pas op: ook rommel).
- `git restore --staged <bestand>` — unstage (haal uit staging), laat je file-inhoud ongemoeid.
- `git commit -m "bericht"` — commit je staged wijzigingen met een message.
- `git commit --amend` — voeg staged wijzigingen toe aan je laatste commit (of pas de message aan).

## 3) Branches
- `git branch` — lijst branches.
- `git branch -vv` — branches + upstream + laatste commit.
- `git switch <branch>` — wissel naar een branch (moderne variant).
- `git switch -c <nieuwenaam>` — maak een nieuwe branch en switch ernaartoe.
- `git checkout <branch>` — oudere variant van switch (kan nog overal werken).
- `git merge <branch>` — merge `<branch>` in je huidige branch.
- `git rebase <branch>` — “verplaats” je huidige branch bovenop `<branch>` (rewrites history: wees voorzichtig).

## 4) Remote & sync (GitHub/GitLab)
- `git remote -v` — toon remote(s) (meestal `origin`).
- `git fetch` — haal updates binnen van remote, zonder je werkbranch te veranderen.
- `git pull` — fetch + integrate (merge of rebase, afhankelijk van config).
- `git push` — push je commits naar remote.
- `git push -u origin <branch>` — eerste push van een nieuwe branch + stel upstream in.

## 5) Tijdelijk parkeren (stash)
- `git stash` — zet lokale wijzigingen tijdelijk weg.
- `git stash pop` — haal de stash terug en verwijder hem uit de stash-lijst.
- `git stash list` — toon alle stashes.
- `git stash apply` — haal stash terug zónder hem te verwijderen.

## 6) Undo/repair (de “oeps”-sectie)
- `git restore <bestand>` — gooi lokale wijzigingen in `<bestand>` weg (unstaged).
- `git restore .` — gooi alle lokale (unstaged) wijzigingen weg.
- `git reset <bestand>` — unstage een bestand (oude manier; vergelijkbaar met `restore --staged`).
- `git reset --soft HEAD~1` — maak laatste commit ongedaan, maar laat changes staged.
- `git reset --mixed HEAD~1` — laatste commit weg, changes blijven als unstaged (default reset).
- `git reset --hard HEAD~1` — laatste commit weg + changes weg (gevaarlijk).
- `git revert <commit>` — maak een “tegen-commit” die een commit ongedaan maakt (veilig voor gedeelde history).
- `git reflog` — je reddingsboei: toont waar HEAD allemaal geweest is (handig na reset/rebase).

## 7) Tags (releases)
- `git tag` — lijst tags.
- `git tag <naam>` — maak een tag op huidige commit.
- `git push --tags` — push tags naar remote.

## Handige “kwaliteit van leven” opties
- `git log --oneline --graph --decorate --all` — mooi overzicht van de geschiedenis/branches.
- `git blame <bestand>` — wie heeft welke regel wanneer aangepast (handig bij bugs).
- `git show <commit>` — toon details van een commit (diff + metadata).

## Mini-werkflow (typisch)
1. `git status`
2. `git add .`  *(of specifieker: `git add <bestanden>`)*  
3. `git commit -m "..."`  
4. `git push`
