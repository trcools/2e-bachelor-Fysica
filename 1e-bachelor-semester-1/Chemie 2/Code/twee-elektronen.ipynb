# Twee elektronen in de oneindige potentiaalput 
## De onafhankelijke elektronbenadering

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# De oplossingen zijn:
a=1
n=2
m=3
def psi_nm(x_1, x_2, length=a,n=n,m=m):
    return (2/length)*np.sin(((n*math.pi)/length)*x_1)*np.sin(((m*math.pi)/length)*x_2)

# Visualisatie plot (positie x_1 in functie van x_2)
def get_data(length=a,n=n,m=m):
    x_1 = np.linspace(0, length, 1000)
    x_2 = np.linspace(0, length , 1000)
    X_1, X_2 = np.meshgrid(x_1, x_2)

    # Bereken de golffunctie over het hele rooster
    Z = psi_nm(X_1, X_2, length=a, n=n, m=m)
    return X_1, X_2, Z



```

```python
X_1, X_2, Z = get_data(length=a,n=n,m=m)

def plot_wavefunction(X_1, X_2, Z, n, m):
    plt.close()
    # oppervlakte plot
    plt.contourf(X_1, X_2, Z, levels=20, cmap='coolwarm')
    plt.xlabel(r'$x_1$')
    plt.ylabel(r'$x_2$')
    plt.title(rf'Golffunctie $\Psi_{{n={n},m={m}}}(x_1, x_2)$')
    plt.colorbar()

    plt.show()
plot_wavefunction(X_1, X_2, Z, n, m)
```

```python
# Symmetrische en antisymmetrische golffuncties
def psi_sym(x_1, x_2, length=a,n=n,m=m):
    return (1/np.sqrt(2))*(psi_nm(x_1, x_2, length=length,n=n,m=m) + psi_nm(x_2, x_1, length=length,n=n,m=m))
def psi_antisym(x_1, x_2, length=a,n=n,m=m):
    return (1/np.sqrt(2))*(psi_nm(x_1, x_2, length=length,n=n,m=m) - psi_nm(x_2, x_1, length=length,n=n,m=m))

def get_data_sym_antisym(length=a,n=n,m=m):
    x_1 = np.linspace(0, length, 1000)
    x_2 = np.linspace(0, length , 1000)
    X_1, X_2 = np.meshgrid(x_1, x_2)

    # Bereken de golffunctie over het hele rooster
    Z_sym = psi_sym(X_1, X_2, length=a, n=n, m=m)
    Z_antisym = psi_antisym(X_1, X_2, length=a, n=n, m=m)
    return X_1, X_2, Z_sym, Z_antisym

def plot_wavefunctions_sym_antisym(X_1, X_2, Z_sym, Z_antisym, n, m):
    plt.close()
    # oppervlakte plot
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    plt.contourf(X_1, X_2, Z_sym, levels=20, cmap='coolwarm')
    plt.xlabel(r'$x_1$')
    plt.ylabel(r'$x_2$')
    plt.title(rf'Symmetrische Golffunctie $\Psi_{{n={n},m={m}}}^{{\mathrm{{sym}}}}(x_1, x_2)$')
    plt.colorbar()

    plt.subplot(1, 2, 2)
    plt.contourf(X_1, X_2, Z_antisym, levels=20, cmap='coolwarm')
    plt.xlabel(r'$x_1$')
    plt.ylabel(r'$x_2$')
    plt.title(rf'Antisymmetrische Golffunctie $\Psi_{{n={n},m={m}}}^{{\mathrm{{antisym}}}}(x_1, x_2)$')
    plt.colorbar()

    plt.tight_layout()
    plt.show()




```

```python
X_1, X_2, Z_sym, Z_antisym = get_data_sym_antisym(length=a,n=n,m=m)
plot_wavefunctions_sym_antisym(X_1, X_2, Z_sym, Z_antisym, n, m)
```
