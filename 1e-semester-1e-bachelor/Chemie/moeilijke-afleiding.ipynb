We vertrekken van de (molaire) vrije enthalpie van menging volgens het roostermodel:

$$
g(x_A)=\frac{z\,w_{AA,m}}{2}\,x_A+\frac{z\,w_{BB,m}}{2}\,x_B
+RT\left(x_A\ln x_A+x_B\ln x_B\right)+RT\,\chi_{AB}\,x_Ax_B
$$

met
$$
x_A=\frac{n_A}{n},\qquad x_B=\frac{n_B}{n}=1-x_A,\qquad n=n_A+n_B.
$$

Het chemisch potentiaal is gedefinieerd via de totale Gibbs-energie:

$$
\mu_A=\left(\frac{\partial G_{\text{tot}}}{\partial n_A}\right)_{T,P,n_B}.
$$

Omdat $g$ per mol is, geldt:

$$
G_{\text{tot}} = n\,g(x_A).
$$

Dus:

$$
\mu_A=\left(\frac{\partial (n g)}{\partial n_A}\right)_{T,P,n_B}
= g + n\frac{dg}{dx_A}\frac{\partial x_A}{\partial n_A}.
$$

Bereken nu $\frac{\partial x_A}{\partial n_A}$ bij vast $n_B$:

$$
x_A=\frac{n_A}{n_A+n_B}=\frac{n_A}{n}
$$

$$
\frac{\partial x_A}{\partial n_A}
=\frac{(n_A+n_B)-n_A}{(n_A+n_B)^2}
=\frac{n_B}{n^2}
=\frac{x_B}{n}.
$$

Invullen geeft de handige algemene identiteit:

$$
\mu_A = g + x_B\frac{dg}{dx_A}.
$$

(Analoog volgt ook: $\,\mu_B = g - x_A\frac{dg}{dx_A}$.)

---

## Afgeleide $dg/dx_A$

Schrijf voor compactheid:

$$
a=\frac{z\,w_{AA,m}}{2},\qquad b=\frac{z\,w_{BB,m}}{2},\qquad x_B=1-x_A.
$$

Dan is

$$
g(x_A)=a x_A+b(1-x_A)+RT\left(x_A\ln x_A+(1-x_A)\ln(1-x_A)\right)+RT\chi_{AB}x_A(1-x_A).
$$

Term per term:

1) Lineaire termen:
$$
\frac{d}{dx_A}\left(ax_A+b(1-x_A)\right)=a-b.
$$

2) Log-termen:
$$
\frac{d}{dx_A}\left(x_A\ln x_A\right)=\ln x_A+1,
$$
$$
\frac{d}{dx_A}\left((1-x_A)\ln(1-x_A)\right)=-(\ln(1-x_A)+1).
$$

Samen:
$$
\frac{d}{dx_A}\left(x_A\ln x_A+x_B\ln x_B\right)
=\ln x_A-\ln x_B
=\ln\left(\frac{x_A}{x_B}\right).
$$

3) Interactieterm:
$$
\frac{d}{dx_A}\left(x_Ax_B\right)
=\frac{d}{dx_A}\left(x_A(1-x_A)\right)
=1-2x_A.
$$

Dus:

$$
\frac{dg}{dx_A}=(a-b)+RT\ln\left(\frac{x_A}{x_B}\right)+RT\chi_{AB}(1-2x_A).
$$

---

## Invullen in $\mu_A=g+x_B g'(x_A)$ en vereenvoudigen

Start:

$$
\mu_A=
\left(ax_A+bx_B\right)
+RT\left(x_A\ln x_A+x_B\ln x_B\right)
+RT\chi_{AB}x_Ax_B
+x_B\left[(a-b)+RT\ln\left(\frac{x_A}{x_B}\right)+RT\chi_{AB}(1-2x_A)\right].
$$

### (i) Lineaire termen
$$
ax_A+bx_B+x_B(a-b)=ax_A+bx_B+ax_B-bx_B=a(x_A+x_B)=a.
$$

### (ii) Log-termen
$$
x_A\ln x_A+x_B\ln x_B+x_B\left(\ln x_A-\ln x_B\right)
=(x_A+x_B)\ln x_A=\ln x_A.
$$

### (iii) $\chi$-termen
$$
x_Ax_B+x_B(1-2x_A)=x_B-x_Ax_B=x_B^2=(1-x_A)^2.
$$

Alles samen:

$$
\mu_A=a+RT\ln x_A+RT\chi_{AB}(1-x_A)^2.
$$

Terug uitschrijven in de oorspronkelijke parameters:

$$
\boxed{\mu_A=\frac{z\,w_{AA,m}}{2}+RT\ln x_A+RT\,\chi_{AB}(1-x_A)^2.}
$$

---

## Analoog voor component $B$

$$
\boxed{\mu_B=\frac{z\,w_{BB,m}}{2}+RT\ln x_B+RT\,\chi_{AB}(1-x_B)^2.}
$$

Omdat $x_B=1-x_A$ kan dit ook als:

$$
\mu_B=\frac{z\,w_{BB,m}}{2}+RT\ln(1-x_A)+RT\,\chi_{AB}x_A^2.
$$
