Where the 45° line intersects the data around $Z_n \approx 40$, that point suggests a periodic orbit at that value of $Z_n$. We call this value $Z^*$.


Is this this $Z^*$ stable?



With a first order taylor expansion and saying that $Z^* \approx f(Z^*)$
we find that $|n_{n+1}|>|n_n|$ so this $Z^*$ is unstable, so the periodic orbit is also unstable.

For larger possible periods, we can use the cobweb method to find them.
$Z^*$ is an example of a period one orbit. But now we want to see if we can find a periodic-p orbit, which comes back to the initial $Z$ value after p iterations.


The other question is if this periodic orbit is stable?

The answer is no, because the magnitude or norm is everywhere greater than one.
The Lorenz map is unstable for any periodic-p. We can conclude from this that the Lorenz attractor is a chaotic attractor.
