# Hoofdstuk 2 — Notatie cheat sheet (Groups & Representations)

## 1) Groep, bewerking, identiteit, inverse
- $G$: een (eindige) groep, met elementen $g,h,x,y,\dots$
- $gh$ (of $g\cdot h$): de groepsbewerking (“vermenigvuldiging”)
- $e$ of $1$: identiteit (neutraal element), met $ge=eg=g$
- $g^{-1}$: inverse van $g$, met $gg^{-1}=g^{-1}g=e$
- $|G|$ of $\#G$: orde/kardinaliteit = aantal elementen in $G$
- $\mathrm{ord}(g)$: orde van een element $g$ = kleinste $n\ge 1$ zodat $g^n=e$

**Abels geval:** men schrijft vaak $x+y$ i.p.v. $xy$, en de identiteit als $0$ i.p.v. $1$.

---

## 2) Bekende groepen (namen die overal terugkomen)
- $\mathbb{Z}_n$: cyclische groep van orde $n$ (integers mod $n$)
- $S_n$: permutatiegroep op $n$ elementen
- $A_n$: even permutaties (subgroep van $S_n$)
- $D_n$: dihedrale groep (symmetrieën van een $n$-hoek)

**Presentatie (generatoren & relaties):**
- $G=\langle S\mid R\rangle$ betekent: $G$ wordt gegenereerd door $S$ met relaties $R$.
- Voorbeeld:
  $$D_n=\langle r,s\mid r^n=e,\ s^2=e,\ (sr)^2=e\rangle$$

---

## 3) Subgroepen
- $H\le G$: $H$ is een subgroep van $G$
- $H<G$: strikte subgroep (dus $H\neq G$)

---

## 4) Cosets en index (waar Lagrange over gaat)
**Linker-coset:**
$$gH=\{gh\mid h\in H\}$$

**Rechter-coset:**
$$Hg=\{hg\mid h\in H\}$$

- $G/H$: de *verzameling* van alle linker-cosets $gH$
- $[G:H]$: de **index** = aantal cosets (dus $[G:H]=|G/H|$)

**Belangrijke waarschuwing:**
- $G/H$ is **niet automatisch een groep**. Dat wordt pas een groep als $H$ *normaal* is.

**Lagrange (idee in 1 zin):**
- Cosets partitioneren $G$ en elk coset heeft $|H|$ elementen, dus
  $$|G|=|H|\,[G:H]$$

---

## 5) Conjugatie, conjugacy classes, normaal
**Conjugatie van $g$ door $h$:**
$$hgh^{-1}$$

**Conjugacy class van $g$:**
$$C_g=\{hgh^{-1}\mid h\in G\}$$

**Normale subgroep:**
- $N\trianglelefteq G$ betekent: $N$ is normaal in $G$
- Intuïtief: “invariant onder conjugatie”
  $$gNg^{-1}=N\quad \text{voor alle } g\in G$$
- Equivalent (handig): $gN=Ng$ voor alle $g\in G$

---

## 6) Quotiëntgroep (factor group)
Als $N\trianglelefteq G$, dan is $G/N$ wél een groep met elementen cosets $gN$ en product:
$$(g_1N)(g_2N)=(g_1g_2)N$$

---

## 7) Centrum en commutatoren
**Centrum:**
$$Z(G)=\{z\in G\mid zg=gz\ \text{voor alle } g\in G\}$$

**Commutator van $x,y$:**
$$[x,y]=xyx^{-1}y^{-1}$$

**Commutatorgroep:**
- $[G,G]$: subgroep gegenereerd door alle commutatoren $[x,y]$
- $G/[G,G]$: “abelianization” (maakt $G$ abels door commutatoren weg te modden)

---

## 8) Homomorfismen en kernels
- $\varphi:G_1\to G_2$: groepshomomorfisme (respecteert de bewerking)
  $$\varphi(xy)=\varphi(x)\varphi(y)$$
- $\ker(\varphi)=\{g\in G_1\mid \varphi(g)=e\}$
- Belangrijk: $\ker(\varphi)\trianglelefteq G_1$ (kernel is altijd normaal)

---

## 9) Automorfismen en semidirect product
- $\mathrm{Aut}(N)$: automorfismengroep van $N$
- $\mathrm{Inn}(G)$: inner automorphisms (conjugatie-automorfismen)

**Semidirect product:**
- Gegeven een actie $\beta:H\to \mathrm{Aut}(N)$, schrijf je:
  $$N\rtimes_\beta H$$
- Elementen zijn paren $(n,h)$ met product:
  $$(n_1,h_1)(n_2,h_2)=(n_1\,\beta_{h_1}(n_2),\ h_1h_2)$$
- Als de actie triviaal is ($\beta_{h}=\mathrm{id}$ voor alle $h$), dan wordt dit het direct product $N\times H$.

---

## Eén zin om het te onthouden
- **Coset** = “$H$ rondschuiven in $G$”
- **Normaal** = “links en rechts rondschuiven komt overeen ($gN=Ng$)”
- **Quotiënt** = “cosets worden de nieuwe ‘elementen’ van een groep”
- **Semidirect** = “twee groepen samenbouwen + een actie die ‘twist’ geeft”
