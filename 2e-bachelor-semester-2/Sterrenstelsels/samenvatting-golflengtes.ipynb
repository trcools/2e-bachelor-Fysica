Waarvoor dienen verschillende golflengten?

Radiogolf + Microgolf (= mm gebied): (>1mm)
	Aardatmosfeer doorzichtig, wel opletten voor menselijke bronnen
	Interferometrie
	Als ellips sterrenstelsel boxy is dan fel in radiogolf
	Totale massa atomair gas sterrenstelsel uit 21 cm lijn van H I halen
	Thermische remstraling (relatief koudere objecten)
	Synchotronstraling (hiermee magnetische velden opsporen en komt voor in buurt van supernova’s en is dus maat voor recente stervorming)
	Stervorminggebied is doorzichtig (zien + energie kwijt)
	Moleculair gas meten via CO molecule is dus tracer voor koude interstellair medium
	Kosmische achtergrond straling (CMB)

Sub mm-gebied= VIR: (30 μm-1mm)
	Moeilijk gebied voor waarnemingen (atmosfeer opaak, koeling nodig, hoge diffractielimiet)
	(CO molecule opnieuw)
	Interstellair  stof zendt dit uit
	Emissie overgang van O V→O VI gebeurt in dit gebied en is een indicator voor een hete galactische corona
	Stervorminggebied is doorzichtig (zien + energie kwijt)
	Starbursts die zoveel stof hebben dat ze vooral submm uitstralen worden SMGs genoemd

IR:  5-30μm
	Makkelijker om in waar te nemen dan submm (door kleinere diffractielimiet + efficiëntere detectoren ontwikkeld) wel nog steeds opake atmosfeer (buiten 2 kleine vensters)
	Normaal Stof zend ook IR straling uit in hogere golflengtes (5-600μm)
	Je hebt ook VSG’s die licht uitstralen in het MIR.

NIR: (750nm -5 μm)
	Atmosfeer niet volledig transparant
	Rode reuzensterren zenden dit licht uit (handig voor stellaire populatie)
	Wordt niet gehinderd door stof
	Neutrale atomaire medium opmeten via absorptielijnen atomair waterstof

Zichtbaar: (400-750nm)
	Atmosfeer is doorlaatbaar, maar last van seeiing
	Neutrale atomaire medium opmeten via absorptielijnen atomair waterstof
	De Ly α lijn ligt in het zichtbare voor zeer hoge roodverschuiving
	Een balmer foton light in het zichtbare
	Veel sterren schijnen vooral licht uit in dit gebied
	Interstellair stof absorbeert zichtbaar licht

UV (10-400nm):
	Vanuit ruimte
	Straling van hete sterren (O/B/A) is UV (goede maat voor SFR)
	UV wordt makkelijk verstrooit/geabsorbeerd door stof
	Neutrale atomaire medium opmeten via absorptielijnen atomair waterstof
	Geïoniseerd gas zendt UV straling uit, Lyman straling
	Ly α lijn is zeer belangrijk om roodverschuiving sterrenstelsels mee te meten
	Goede manier om sterrenstelsels met een zeer opmerkelijk hoge SFR te vinden

Röntgen: (0.01nm-10nm)
	Ruimte waarnemingen + grazing incidence spiegels nodig
	Boxy elliptische sterrenstelsels zijn lichtkrachtiger in Röntgengebied
	Remstraling ligt in Röntgen voor zeer hete voorwerpen
	Accretie nabij quasar
	Bij supernova’s komt er X-ray straling vrijdag
	Heet geïoniseerd gas zend dit uit
	Rond sterrenstelsels is er een halo van heet geïoniseerd gas dat Röntgenstraling uitzend (handig voor massa ifv straal te achterhalen)


Gamma: (<0.01nm)
	Om massa moleculaire gaswolken te achterhalen (kosmische straling interageert met interstellair gas deeltje met gamma straling als gevolg)
